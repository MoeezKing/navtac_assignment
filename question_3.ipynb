```python
import matplotlib.pyplot as pyp
import pandas as pd
```

```python
# univariate plot
std_marks = pd.DataFrame([33,45,67,78,33,67,44,33,80,90])
up = pyp.hist(std_marks)
pyp.show()
```

```python
# bivariate plot
x = pd.DataFrame([2,3,4,5,6,8])
y = pd.DataFrame([10,5,15,18,20,7])
bp = pyp.scatter(x, y , color='yellow')
pyp.xlabel('Number of Employee')
pyp.ylabel('Sales')
pyp.show()
```

```python
# multivariate plot
x = pd.DataFrame([1,2,3,4,5])
y = pd.DataFrame([5,4,3,2,1])
z = pd.DataFrame([0,1,5,4,3])

mp1 = pyp.plot(x,y, color='r')
mp2 = pyp.plot(y,z, color='b')
pyp.show()
```
