```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import pandas as pd
import matplotlib.pyplot as pyp
```

```python
# reading csv 
data = pd.read_csv('house_data.csv')
```

```python
# take attributes as input and target as output
attributes = ['bedrooms','bathrooms']
target = 'price'
```

```python
# getting data from csv as input (X) and output (Y)
X = data[attributes]
Y = data[target]
```

```python
# now split the X,Y tests and splits
x_train,x_test,y_train,y_test = train_test_split(X,Y)

print(f'x-train :{x_train} \nand x-test : {x_test}')
print(f'y-train :{y_train} \nand y-test : {y_test}')
```

```python
# defining a model for linear regression
model = LinearRegression()
```

```python
# training the model
model.fit(x_train, y_train)
```

```python
house = [[3,2]]
prediction = model.predict(house)
print(prediction)
```

```python
sp1 = pyp.scatter(X['bedrooms'],Y)
sp2 = pyp.scatter(house[0][1],prediction, color='red')
pyp.xlabel('Bedrooms')
pyp.ylabel('Prices')
# sp2 = pyp.scatter(Y,X['bathrooms'])
pyp.show()
```
